```python
import pandas as pd

# Create a small dictionary of data
data = {
    'Epoch': [1, 2, 3, 4, 5],
    'Loss': [0.95, 0.72, 0.45, 0.21, 0.12],
    'Accuracy': [0.55, 0.68, 0.82, 0.91, 0.94]
}

# Save it as a CSV
df_to_save = pd.DataFrame(data)
df_to_save.to_csv('test_data.csv', index=False)
print("File 'test_data.csv' created successfully!")
```

```python
import matplotlib.pyplot as plt

# Step A: Load the CSV
df = pd.read_csv('test_data.csv')

# Step B: Display the DataFrame (This renders the nice HTML table)
display(df)

# Step C: Plot a chart
plt.figure(figsize=(8, 4))
plt.plot(df['Epoch'], df['Loss'], label='Training Loss', marker='.')
plt.title('Model Training Progress')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```
